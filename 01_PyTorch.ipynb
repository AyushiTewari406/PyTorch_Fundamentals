Add an extra dimension

Tensor Reshaping Summary
Reshaping means changing the shape/dimensions of a tensor without changing its data.

Example:

x = torch.tensor([1,2,3,4,5,6])

x.reshape(2,3)

Output:

[[1,2,3],
 [4,5,6]]
Key Rule

Total elements must remain same.

Valid:

6 → (2×3), (3×2), (1×6)

Invalid:

(4×2) = 8 elements

```python
x_reshaped=x.reshape(3,3)
x_reshaped,x_reshaped.shape
```

```python
x_reshaped=x.reshape(9,1)
x_reshaped,x_reshaped.shape
```

Change the view

Both are used to change tensor shape.

| `view()`                         | `reshape()`                                         |
| -------------------------------- | --------------------------------------------------- |
| Only works on contiguous tensors | Works on both contiguous and non-contiguous tensors |
| Faster                           | Slightly safer/flexible                             |
| May give error                   | Usually works                                       |
| Does not copy data               | May copy data if needed                             |

permute() is used to rearrange the order of dimensions of a tensor.

Indexing

PyTorch indexing same as NumPy indexing

x[dim0, dim1, dim2]

(1, 3, 3)
Meaning:

1 block
3 rows
3 columns

Meaning of :
x[:, :, 1]

Read it dimension by dimension:

x[ :, :, 1 ]
Dimension 0 → :

Take all values from dim 0.

There is only one block:

[
  [[1,2,3],
   [4,5,6],
   [7,8,9]]
]
Dimension 1 → :

Take all rows

[ [1,2,3],
  [4,5,6],
  [7,8,9] ]
Dimension 2 → 1

Take index 1 from every row.

Remember indexing starts at 0:

0 -> first
1 -> second
2 -> third

So from:

[1,2,3] -> 2
[4,5,6] -> 5
[7,8,9] -> 8

Result:

tensor([[2,5,8]])

Reproducibility

random out of random

In short how a neural network learns
image.png

to reduce the randomness in neural networks and PyTorch comes the concept of random seed

Essentially random seed flavours the randomness

```python
torch.rand(3,3)
```

```python
import torch
random_tensor_A = torch.rand(3,4)
random_tensor_B = torch.rand(3,4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

let's make random reproducible tensors

call manual_seed everytime before tensors

```python
import torch
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)
random_tensor_C = torch.rand(3,4)
torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3,4)
print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C==random_tensor_D)
```
